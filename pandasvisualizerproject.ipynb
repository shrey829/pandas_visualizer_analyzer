```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


print("enter your choices")
print("1.load data")
print("2.explore data")
print("3.perform dataframe operations")
print("4.handling missing values")
print("5.generate descriptive info")
print("6.data visualization")
print("7.save visualization")
print("8.exit")
print("please choose your choice")


class SalesDataAnalyzer:

    def __init__(self, data=None):
        self.data = pd.DataFrame()

    def __del__(self):
        print("deleting the object")


    # --------------------------------------------------
    # LOAD DATA
    # --------------------------------------------------

    def __loaddata(self):

        try:
            print("Give me the CSV file path")
            file_path = str(input())

            self.data = pd.read_csv(file_path)

            print("Data loaded successfully")

        except Exception as e:
            print("Error is:", e)

    # --------------------------------------------------
    # EXPLORE DATA
    # --------------------------------------------------

    def __explore_data(self):

        assert not self.data.empty, "Your DataFrame is empty"

        print("\nData Exploration:")
        print("1. Display first 5 rows")
        print("2. Display last 5 rows")
        print("3. Display column names")
        print("4. Display data types")
        print("5. Display basic info and statistics")
        print("6. Exit")

        c = int(input("Enter your choice: "))

        if c == 1:
            print(self.data.head())

        elif c == 2:
            print(self.data.tail())

        elif c == 3:
            print(self.data.columns)

        elif c == 4:
            print(self.data.dtypes)

        elif c == 5:
            self.data.info()
            print(self.data.describe())

        elif c == 6:
            exit()

        else:
            print("Invalid choice")

    # --------------------------------------------------
    # HANDLING MISSING VALUES
    # --------------------------------------------------

    def __handling_missing_values(self):

        assert not self.data.empty, "Your DataFrame is empty"

        print("\nHandling missing values:")
        print("1. Display rows with missing values")
        print("2. Fill missing values with mean")
        print("3. Drop rows with missing values")
        print("4. Replace missing values with a specific value")
        print("5. Exit")

        d = int(input("Enter your choice: "))

        if d == 1:

            print(
                self.data[
                    self.data.isnull().any(axis=1)
                ]
            )

        elif d == 2:

            self.data.fillna(
                self.data.mean(numeric_only=True),
                inplace=True
            )

            print("Missing numerical values filled with mean")

        elif d == 3:

            self.data.dropna(inplace=True)

            print("Rows with missing values dropped")

        elif d == 4:

            value = input(
                "Enter the value to replace missing values: "
            )

            self.data.fillna(value, inplace=True)

            print("Missing values replaced")

        elif d == 5:
            exit()

        else:
            print("Invalid choice")

    # --------------------------------------------------
    # CONVERSION OPERATION
    # --------------------------------------------------

    def __conversion_operation(self):

        assert not self.data.empty, "Your DataFrame is empty"

        print("\n--- CONVERSION OPERATION ---")

        print("Available columns:")
        print(self.data.columns)

        self.col_name = input("Enter column name: ")

        self.arr = self.data[self.col_name].to_numpy()

        print("\nNumPy Array:")
        print(self.arr)

    # --------------------------------------------------
    # INDEXING OPERATION
    # --------------------------------------------------

    def __indexing_operation(self):

        print("\n--- INDEXING OPERATION ---")

        print("Array:")
        print(self.arr)

        index = int(input("Enter index: "))

        print(
            "Element at index",
            index,
            ":",
            self.arr[index]
        )

    # --------------------------------------------------
    # SLICING OPERATION
    # --------------------------------------------------

    def __slicing_operation(self):

        print("\n--- SLICING OPERATION ---")

        print("1. Slicing whole dataframe")
        print("2. Slicing specific column")
        print("3. Slicing specific row")

        choice = int(input("Enter choice: "))

        # ----------------------------------------------
        # WHOLE DATAFRAME
        # ----------------------------------------------

        if choice == 1:

            print("\n--- WHOLE DATAFRAME SLICING ---")

            rs = int(input("Row start: "))
            re = int(input("Row end: "))

            cs = int(input("Column start: "))
            ce = int(input("Column end: "))

            print("\nSliced DataFrame:")

            print(
                self.data.iloc[
                    rs:re,
                    cs:ce
                ]
            )

        # ----------------------------------------------
        # SPECIFIC COLUMN
        # ----------------------------------------------

        elif choice == 2:

            print("\n--- SPECIFIC COLUMN SLICING ---")

            start = int(input("Column start: "))
            end = int(input("Column end: "))

            print("\nSliced Array:")

            print(
                self.arr[start:end]
            )

        # ----------------------------------------------
        # SPECIFIC ROW
        # ----------------------------------------------

        elif choice == 3:

            print("\n--- SPECIFIC ROW SLICING ---")

            row = int(input("Enter row index: "))

            arr = self.data.iloc[row].to_numpy()

            print("Selected row:")
            print(arr)

            start = int(input("Row start: "))
            end = int(input("Row end: "))

            print("\nSliced Row:")

            print(
                arr[start:end]
            )

        else:
            print("Invalid choice")

    # --------------------------------------------------
    # INDEXING / SLICING / CONVERSION MENU
    # --------------------------------------------------

    def __indexing_slicing_conversion_operations(self):

        assert not self.data.empty, "Your DataFrame is empty"

        print("\n--- INDEXING / SLICING / CONVERSION ---")

        print("1. Conversion")
        print("2. Indexing")
        print("3. Slicing")

        choice = int(input("Enter choice: "))

        if choice == 1:

            self.__conversion_operation()

        elif choice == 2:

            self.__conversion_operation()
            self.__indexing_operation()

        elif choice == 3:

            self.__conversion_operation()
            self.__slicing_operation()

        else:

            print("Invalid choice")

    # --------------------------------------------------
    # MATHEMATICAL OPERATION
    # --------------------------------------------------

    def __mathematical_operation(self):

        assert not self.data.empty, "Your DataFrame is empty"

        print("\n--- MATHEMATICAL OPERATIONS ---")

        print("1.sum of columns of dataframes")
        print("2.mean of columns of dataframes")
        print("3.median of columns of dataframes")
        print("4.mode of columns of dataframes")
        print("5.exit")
        print("enter your choice")
        g = int(input())
        if g == 1:
            print("sum of columns of dataframes")
            print(self.arr.sum())

        elif g == 2:
            print("mean of columns of dataframes")
            print(self.arr.mean())
        elif g == 3:
            print("median of columns of dataframes")
            print(np.median(self.arr))
        elif g == 4:
            print("mode of columns of dataframes")
            print(pd.Series(self.arr).mode())
        else:
            print("INVALID CHOICE")

    def __combine(self):
        assert not self.data.empty, "Your DataFrame is empty"

        file_path1 = input("enter the file path of  dataframe: ")
        df1 = pd.read_csv(file_path1)

        print("1.concatination of dataframes")
        print("2.join of dataframes")
        print("3.merge of dataframes")

        d = int(input("enter your choice: "))
        if d == 1:
            print("concatination of dataframes")
            df2 = pd.concat([self.data, df1], ignore_index=True)
            print(df2)
        elif d == 2:
            print("join of dataframes")
            df2 = self.data.join(df1.set_index('id'), on='id', how='inner')
            print(df2)
        elif d == 3:
            print("merge of dataframes")
            df2 = pd.merge(self.data, df1, on="id", how="inner")
            print(df2)

        else:
            print("invalid choice")

    def __split(self):
        assert not self.data.empty, "Your DataFrame is empty"

        print("1.split dataframe based on regions")
        print("2.split dataframe based on products")

        print("4.split dataframe based on your choice of criteria")

        d = int(input("enter your choice for operations: "))
        if d == 1:
            print("split dataframe based on regions")
            print("give me the columns name for the region columns in your dataframe ")
            colname = str(input(""))
            lst = []

            df1 = self.data[colname].unique()
            for i in df1:
                lst.append(self.data[self.data[colname] == i])
            for i in lst:
                print(i)

        elif d == 2:
            print("split dataframe based on products")
            print("give me the columns name for the product columns in your dataframe")
            colname = str(input(""))
            lst = []
            df1 = self.data[colname].unique()
            for i in df1:
                lst.append(self.data[self.data[colname] == i])
            for i in lst:
                print(i)

        elif d == 4:
            print("split dataframe based on your choice of criteria")
            print("give me the columns name for the criteria columns in your dataframe")
            colname = str(input(""))
            lst = []
            df1 = self.data[colname].unique()
            for i in df1:
                lst.append(self.data[self.data[colname] == i])
            for i in lst:
                print(i)
        else:
            print("invalid choice")

    def __search(self):
        assert not self.data.empty, "Your DataFrame is empty"

        print("1.search by column value")
        print("2.search by row index")

        i = int(input())
        if i == 1:
            print("search by column value")
            colname = input("enter the column name: ")
            value = input("enter the value to search: ")
            print(self.data[self.data[colname].astype(str) == value])
        elif i == 2:
            print("search by row index")
            index = int(input("enter the row index: "))
            print(self.data.loc[index])
        else:
            print("Invalid choice")

    def __sort(self):
        assert not self.data.empty, "Your DataFrame is empty"

        print("1.sort by column value")
        print("2.sort by row index")

        j = int(input())
        if j == 1:
            print("sort by column value")
            colname = input("enter the column name: ")
            print(self.data.sort_values(by=colname))
        elif j == 2:
            print("sort by row index")
            print(self.data.sort_index())

        else:
            print("invalid choice")

    def __filter(self):
        assert not self.data.empty, "Your DataFrame is empty"

        print("1.filter by column value")
        print("2.filter by row index")

        k = int(input())
        if k == 1:
            print("filter by column value")
            colname = input("enter the column name: ")
            value = input("enter the value to filter: ")
            print(self.data[self.data[colname].astype(str) == value])
        elif k == 2:
            print("filter by row index")
            index = int(input("enter the row index: "))
            print(self.data.loc[index])

        else:
            print("invalid choice")

    def __aggregate(self):
        assert not self.data.empty, "Your DataFrame is empty"

        print("give us the columns name for the aggregation columns in your dataframe")
        a = str(input(""))
        print("1.sum of columns of dataframes")
        print("2.mean of columns of dataframes")
        print("3.median of columns of dataframes")
        print("4.mode of columns of dataframes")
        print("5.exit")
        print("enter your choice")
        l = int(input())
        if l == 1:
            print("sum of columns of dataframes")
            print(self.data.groupby(a).sum(numeric_only=True))
        elif l == 2:
            print("mean of columns of dataframes")
            print(self.data.groupby(a).mean(numeric_only=True))
        elif l == 3:
            print("median of columns of dataframes")
            print(self.data.groupby(a).median(numeric_only=True))
        elif l == 4:
            print("mode of columns of dataframes")
            print(self.data.groupby(a).agg(lambda x: x.mode().iloc[0]))
        elif l == 5:
            exit()
        else:
            print("invalid choice")

    def __statistical_analysis(self):
        assert not self.data.empty, "Your DataFrame is empty"

        print('give me the columns name for the descriptive statistics columns in your dataframe and the values for the descriptive statistics columns in your dataframe')
        a = str(input(""))
        b = str(input(""))
        print(self.data.groupby(a)[b].agg(['mean', 'median', lambda x: x.mode().iloc[0], 'std', 'var']))
        print(self.data[b].quantile([0.25, 0.75]))
        print(self.data.describe())
        self.data.info()

    def __show_plot(self, file_path=None):
        if file_path:
            self.__save_plot(file_path)
        plt.show()

    def __matplotlib_data_visualizer(self, file_path=None):
        print("1.line plot")
        print("2.bar plot")
        print("3.scatter plot")
        print("4.histogram")
        print("6.pie chart")

        print("7.stack plots")
        print("8.subplots")

        print("enter your choice")
        o = int(input())
        if o == 1:
            print("line plot")
            print("give me the columns name for the x axis and y axis in your dataframe")
            x = str(input(""))
            y = str(input(""))
            plt.figure(figsize=(12, 6))

            plt.plot(
                self.data[x],
                self.data[y],
                marker="o",
                linewidth=2,
                label="Sales"
            )

            plt.xlabel(x)
            plt.ylabel(y)
            plt.xticks(rotation=45)
            plt.grid(True, alpha=0.3)
            plt.legend()

            plt.tight_layout()
            self.__show_plot(file_path)
        elif o == 2:
            print("bar plot")
            print("give me the columns name for the x axis and y axis in your dataframe")
            x = str(input(""))
            y = str(input(""))
            plt.figure(figsize=(12, 6))
            plt.bar(
                self.data[x],
                self.data[y],
                color="skyblue",
            )

            plt.xlabel(x)
            plt.ylabel(y)
            plt.xticks(rotation=45)
            plt.grid(True, alpha=0.3)

            plt.tight_layout()
            self.__show_plot(file_path)
        elif o == 3:
            print("scatter plot")
            print("give me the columns name for the x axis and y axis in your dataframe")
            x = str(input(""))
            y = str(input(""))
            plt.figure(figsize=(12, 6))
            plt.scatter(
                self.data[x],
                self.data[y],
                color="green",
                marker="o",
            )
            plt.xlabel(x)
            plt.ylabel(y)
            plt.xticks(rotation=45)
            plt.grid(True, alpha=0.3)
            plt.tight_layout()
            self.__show_plot(file_path)
        elif o == 4:
            print("histogram")
            print("give me the columns name for the x axis in your dataframe")
            x = str(input(""))
            plt.figure(figsize=(12, 6))
            plt.hist(
                self.data[x],
                bins=10,
                color="orange",
                alpha=0.7,
            )
            plt.xlabel(x)
            plt.ylabel("Frequency")
            plt.xticks(rotation=45)
            plt.grid(True, alpha=0.3)
            plt.tight_layout()
            self.__show_plot(file_path)
        elif o == 6:
            print("pie chart")
            print("give me the columns name for the x axis and y axis in your dataframe")
            x = str(input(""))
            y = str(input(""))
            plt.figure(figsize=(8, 8))
            plt.pie(
                self.data[y],
                labels=self.data[x],
                autopct="%1.1f%%",
                startangle=90,
            )
            plt.title("Pie Chart")
            plt.axis("equal")
            plt.tight_layout()
            self.__show_plot(file_path)
        elif o == 7:
            print("stack plots")
            print("give me the columns name for the x axis and y axis in your dataframe")
            x = str(input(""))
            y = str(input(""))
            plt.figure(figsize=(12, 6))
            plt.stackplot(
                self.data[x],
                self.data[y],
                labels=[y],
                alpha=0.7,
            )
            plt.xlabel(x)
            plt.ylabel(y)
            plt.xticks(rotation=45)
            plt.grid(True, alpha=0.3)
            plt.legend()
            plt.tight_layout()
            self.__show_plot(file_path)
        elif o == 8:
            print("subplots")
            print("Give me the column names for the x-axis and y-axis for both plots")

            x = str(input("Enter x-axis column for plot 1: "))
            y = str(input("Enter y-axis column for plot 1: "))
            x2 = str(input("Enter x-axis column for plot 2: "))
            y2 = str(input("Enter y-axis column for plot 2: "))

            plt.figure(figsize=(12, 10))

            # First subplot
            plt.subplot(2, 1, 1)
            plt.plot(self.data[x], self.data[y], marker="o", color="blue")
            plt.title("Line Plot 1")
            plt.xlabel(x)
            plt.ylabel(y)

            # Second subplot
            plt.subplot(2, 1, 2)
            plt.plot(self.data[x2], self.data[y2], marker="s", color="red")
            plt.title("Line Plot 2")
            plt.xlabel(x2)
            plt.ylabel(y2)

            plt.tight_layout()
            self.__show_plot(file_path)

        else:
            print("invalid chocie")

    def __seaborn_data_visualizer(self, file_path=None):
        print("1.heatmap")
        print("2.boxplot")
        print("3.countplot")
        choice = int(input())
        if choice == 1:
            print("heatmap")
            print("give me the columns name for the x axis and y axis in your dataframe")
            x = str(input(""))
            y = str(input(""))
            plt.figure(figsize=(10, 6))
            corr = self.data.pivot_table(index=x, columns=y, aggfunc="size", fill_value=0).corr()
            sns.heatmap(
                corr,
                annot=True,
                fmt=".2f",
                cmap="YlGnBu",
            )
            plt.title("Heatmap")
            plt.tight_layout()
            self.__show_plot(file_path)
        elif choice == 2:
            print("boxplot")
            print("give me the columns name for the x axis and y axis in your dataframe")
            x = str(input(""))
            y = str(input(""))
            plt.figure(figsize=(10, 6))
            sns.boxplot(
                x=self.data[x],
                y=self.data[y],
                hue=self.data[x],
                palette="Set2",
                legend=False,
            )
            plt.title("Boxplot")
            plt.tight_layout()
            self.__show_plot(file_path)
        elif choice == 3:
            print("countplot")
            print("give me the columns name for the x axis in your dataframe")
            x = str(input(""))
            plt.figure(figsize=(10, 6))
            sns.countplot(
                x=self.data[x],
                hue=self.data[x],
                palette="Set3",
                legend=False,
            )
            plt.title("Countplot")
            plt.tight_layout()
            self.__show_plot(file_path)
        else:
            print("invalid choice")

    def __save_plot(self, file_path):
        plt.savefig(file_path)
        print("Plot saved successfully!")

    def __save_visualizer(self, file_path):
        assert not self.data.empty, "Your DataFrame is empty"

        print("1.matplotlib visuals")
        print("2.seaborn visuals")
        choice = int(input())
        if choice == 1:
            print("it should end with png")
            self.__matplotlib_data_visualizer(file_path)
            self.__save_plot(file_path)

        elif choice == 2:
            print("it should end with .png")
            self.__seaborn_data_visualizer(file_path)
            self.__save_plot(file_path)
        else:
            print("invalid choice")

    def main(self):
        while True:
            
                choice = int(input("enter the choice"))

                match choice:
                    case 1:
                        self.__loaddata()
                    case 2:
                        self.__explore_data()
                    case 4:
                        self.__handling_missing_values()
                    case 3:
                        assert not self.data.empty, "Your DataFrame is empty"

                        print("1.convert_index_slicing")
                        print("2.mathematical_combine_split")
                        print("3.search_sort_filter")
                        choice = int(input())
                        if choice == 1:
                            print("1.convert")
                            print("2.index")
                            print("3.slice")
                            choice = int(input())
                            if choice == 1:
                                self.__conversion_operation()
                            elif choice == 2:
                                self.__conversion_operation()
                                self.__indexing_operation()
                            elif choice == 3:
                                self.__conversion_operation()
                                self.__slicing_operation()
                            else:
                                print("invalid choice")
                        elif choice == 2:
                            print("1.mathematical operation")
                            print("2.combine operations")
                            print("3.split operations")
                            choice = int(input())
                            if choice == 1:
                                self.__conversion_operation()
                                self.__mathematical_operation()
                            elif choice == 2:
                                self.__combine()
                            elif choice == 3:
                                self.__split()
                            else:
                                print("invalid choice")
                        elif choice == 3:
                            print("1.search")
                            print("2.sort")
                            print("3.filter")
                            print("4.aggregate")
                            choice = int(input("enter the choice"))
                            if choice == 1:
                                self.__search()
                            elif choice == 2:
                                self.__sort()
                            elif choice == 3:
                                self.__filter()
                            elif choice == 4:
                                self.__aggregate()
                            else:
                                print("INVALID CHOICE")
                    case 5:
                        self.__statistical_analysis()
                    case 6:
                        assert not self.data.empty, "Your DataFrame is empty"

                        print("using matplotlib options are:-[bar plot,line plot,histogram,scatter plot,stack plot,pie chart]")
                        print("using seaborn options are:-[heatmap,boxplot,countplot]")
                        choice = int(input("1.matplotlib,2.Seaborn"))
                        if choice == 1:
                            self.__matplotlib_data_visualizer()
                        elif choice == 2:
                            self.__seaborn_data_visualizer()
                        else:
                            print("invalid choice")
                    case 7:
                        print("enter the file path")
                        file_path = str(input())
                        self.__save_visualizer(file_path)
                    case 8:
                        print("exiting the program")
                        exit()
                    case _:
                        print("invalid choice")


if __name__ == "__main__":
    d = SalesDataAnalyzer()
    d.main()
```
